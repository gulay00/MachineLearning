```python
q1,q3=np.percentile(data['Age'].fillna(data['Age'].mean()),[25,75])
iqr=q3-q1 
lower=q1-1.5 * iqr 
upper=q3+1.5 * iqr 
data=data[(data['Age']>lower) & (data['Age']<upper)] 
```

```python
from sklearn.cluster import KMeans
```

```python
X=data[['Age','Book-Rating','Year-Of-Publication']]
X
```

```python
X.info()
```

```python
X['Year-Of-Publication'] = pd.to_numeric(X['Year-Of-Publication'], errors='coerce')
```

```python
X = X.dropna(subset=['Year-Of-Publication'])
```

```python
from sklearn.preprocessing import StandardScaler

std=StandardScaler()
X=std.fit_transform(X)
```

```python

import matplotlib.pyplot as plt

inertia = []

k_values = range(1,11)

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)   
    inertia.append(kmeans.inertia_)

plt.plot(k_values, inertia, marker='o')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
kmeans=KMeans(n_clusters=5,random_state=42)
data['Cluster']=kmeans.fit_predict(X)
```

```python
data.head()
```

```python
def recommend(book_title, top_n=10):
    recs = data[data['Book-Title'].str.contains(book_title, case=False, na=False)].copy()
    if recs.empty:
        return []
    recs = recs.sort_values('Book-Rating', ascending=False)
    recs = recs['Book-Title'].drop_duplicates()
    return recs.head(top_n)
```
